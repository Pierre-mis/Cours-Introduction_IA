```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

RANDOM_STATE = 42
pd.set_option('display.max_columns', 60)
```

## 1. Chargement


```python
from pathlib import Path

DATA_PATH = Path('data/high_diamond_ranked_10min.csv')

if not DATA_PATH.exists():
    raise FileNotFoundError(
        f"Fichier introuvable : {DATA_PATH.resolve()}\n"
        "Télécharge le CSV depuis :\n"
        "  https://www.kaggle.com/datasets/bobbyscience/league-of-legends-diamond-ranked-games-10-min\n"
        "et place-le dans le sous-dossier data/ de ce TP."
    )

df = pd.read_csv(DATA_PATH)
print(f'Shape : {df.shape}')
df.head()
```

## 2. Sanity check


```python
print('--- dtypes ---')
print(df.dtypes.value_counts())
print('\n--- valeurs manquantes (par colonne, top 10) ---')
print(df.isna().sum().sort_values(ascending=False).head(10))
print('\n--- doublons ---')
print(f'{df.duplicated().sum()} lignes dupliquées')
```

## 3. Target : `blueWins`


```python
target = 'blueWins'
print(df[target].value_counts(normalize=True).rename('proportion'))

fig, ax = plt.subplots(figsize=(4, 3))
df[target].value_counts().plot(kind='bar', ax=ax, color=['#264653', '#e76f51'])
ax.set_title('Distribution de blueWins')
ax.set_xticklabels(['défaite', 'victoire'], rotation=0)
plt.show()
```
