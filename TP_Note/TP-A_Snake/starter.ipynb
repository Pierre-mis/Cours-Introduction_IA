## 1. Prise en main

L'env renvoie un **5-tuple** : `(danger_gauche, danger_devant, danger_droite, dir_pomme_x, dir_pomme_y)`.

- `danger_*` ∈ {0, 1} : mur ou queue dans cette direction *relative* au serpent.
- `dir_pomme_*` ∈ {-1, 0, +1} : direction de la pomme *dans le repère* du serpent.

Actions : **0 = tourner gauche · 1 = tout droit · 2 = tourner droite** (relatif à l'orientation courante).


````python
env = SnakeEnv(grid=8, seed=42)
print('état initial :', env.reset())
print('après step(1):', env.step(1))   # state, reward, done

fig, ax = plt.subplots(figsize=(4, 4))
env.render(ax=ax); plt.show()
````

### À toi (réflexion notée — étape 2 du sujet)

Le 5-tuple **n'encode ni** la position absolue du serpent, **ni** sa longueur, **ni** où est la pomme exactement. Pourtant, l'agent va devoir gérer une pomme qui se téléporte et un serpent qui grandit à chaque pomme.

- Pourquoi cette représentation suffit-elle malgré ces variations ?
- Que perdrait-on (ou que gagnerait-on) à ajouter la position absolue, ou les positions case par case de la queue ?

## 2. Baseline aléatoire (ton point de comparaison)

Avant de coder Q-Learning, mesure ce que fait un agent qui joue au hasard. Ton agent entraîné devra faire **clairement mieux**.


````python
rng = np.random.default_rng(0)
lengths = []
for seed in range(200):
    e = SnakeEnv(grid=8, seed=seed); e.reset()
    while not e.done:
        e.step(int(rng.integers(0, 3)))
    lengths.append(e.length)

print(f'longueur moyenne (200 ép. random) : {np.mean(lengths):.2f}')
````
