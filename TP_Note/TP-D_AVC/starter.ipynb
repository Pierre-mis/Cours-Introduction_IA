**Starter notebook.** Vérifie que les données se chargent et donne un premier aperçu. Sujet complet dans `TP_Note_Sujets.md` (§ TP-D) et grille dans `BAREME.md`.

**Contexte clinique** : un faux négatif (rater un patient à risque) est plus grave qu'un faux positif. Garde ça en tête pour le choix de métriques **et** pour l'ajustement du seuil.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

RANDOM_STATE = 42
```

## 1. Chargement


```python
from pathlib import Path

DATA_PATH = Path('data/healthcare-dataset-stroke-data.csv')

if not DATA_PATH.exists():
    raise FileNotFoundError(
        f"Fichier introuvable : {DATA_PATH.resolve()}\n"
        "Télécharge le CSV depuis :\n"
        "  https://www.kaggle.com/datasets/fedesoriano/stroke-prediction-dataset\n"
        "et place-le dans le sous-dossier data/ de ce TP."
    )

df = pd.read_csv(DATA_PATH)
print(f'Shape : {df.shape}')
df.head()
```

## 2. Sanity check — types et valeurs manquantes


```python
print('--- dtypes ---')
print(df.dtypes)
print('\n--- valeurs manquantes ---')
print(df.isna().sum())
print('\n--- modalités catégorielles ---')
for col in df.select_dtypes(include='object').columns:
    print(f'  {col}: {sorted(df[col].dropna().unique())}')
```

## 3. Target : `stroke` — attention au **fort déséquilibre**


```python
target = 'stroke'
print(df[target].value_counts())
print()
print(df[target].value_counts(normalize=True).rename('proportion'))

fig, ax = plt.subplots(figsize=(4, 3))
df[target].value_counts().plot(kind='bar', ax=ax, color=['#264653', '#e76f51'])
ax.set_title('Distribution de stroke')
ax.set_xticklabels(['pas d\'AVC', 'AVC'], rotation=0)
plt.show()
```

## 4. Baseline triviale

Quelle accuracy obtient un modèle qui prédit toujours 0 ? Pourquoi cette accuracy est-elle trompeuse dans ce contexte ?


```python
baseline_acc = (df[target] == 0).mean()
print(f'Accuracy d\'un classifieur "toujours 0" : {baseline_acc:.3%}')
```
